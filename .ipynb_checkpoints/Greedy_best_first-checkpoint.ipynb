```python
import heapq

def greedy_best_first_search(graph, start, goal, heuristic):
    # Priority queue stores (heuristic value, node, path)
    priority_queue = [(heuristic[start], start, [start])]
    visited = set()

    while priority_queue:
        h_val, node, path = heapq.heappop(priority_queue)

        if node == goal:
            return path

        if node in visited:
            continue

        visited.add(node)

        for neighbor, _ in graph[node]:
            if neighbor not in visited:
                heapq.heappush(priority_queue, (heuristic[neighbor], neighbor, path + [neighbor]))

    return None


# Example graph (same as UCS example)
graph = {
    'A': [('B',1), ('C',3), ('D',1)],
    'B': [('E',10)],
    'C': [('F',8)],
    'D': [('G',4)],
    'E': [],
    'F': [],
    'G': [('H',1)],
    'H': []
}

# Heuristic values (estimated distance to goal 'H')
heuristic = {
    'A': 7,
    'B': 6,
    'C': 5,
    'D': 3,
    'E': 100,
    'F': 100,
    'G': 1,
    'H': 0
}

path = greedy_best_first_search(graph, 'A', 'H', heuristic)
print("Path found by Greedy Best-First Search:", path)
```
