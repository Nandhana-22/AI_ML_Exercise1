```python
 # priority queue module
import heapq

def ucs(graph,start,goal):
    priority_queue=[(0,start,[start])]  #cost,current node,path
    
    #Dictionary to store node and minimum cost
    visited={}
    
    while priority_queue:
        
        #Remove the element with smallest cost
        cost,node,path=heapq.heappop(priority_queue)

        if node==goal:
            return path,cost

        #Skip higher cost visit
        if node in visited and visited[node]<=cost:
            continue

        visited[node]=cost

        for n,w in graph[node]:
            heapq.heappush(priority_queue, (cost+w, n, path+[n]))
    return None

graph= {
    'A': [('B',1), ('C',3), ('D',1)],
    'B': [('E',10)],
    'C': [('F',8)],
    'D': [('G',4)],
    'E': [],
    'F': [],
    'G': [('H',1)]
}

path,cost = ucs(graph, 'A', 'H')
print("Minimum cost:",cost)
print("Path to source to goal:",path)
            
        
```

```python
import heapq
def ucs(graph,start,goal):
    pq=[(0,start,[start])]
    v={}
    while pq:
        cost,node,path=heapq.heappop(pq)
        if node==goal:
            return path,cost
        if node in v and v[node]<=cost:
            continue
        v[node]=cost

        for n,w in graph[node]:
            heapq.heappush(pq, (cost+w, n, (path+[n])))
    return None


graph= {
    'A': [('B',1), ('C',3), ('D',1)],
    'B': [('E',10)],
    'C': [('F',8)],
    'D': [('G',4)],
    'E': [],
    'F': [],
    'G': [('H',1)]
}
path,cost=ucs(graph,'A','H')
print("Minimum cost:",cost)
print("Path to source to goal:",path)
```
