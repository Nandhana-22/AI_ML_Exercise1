```python
import heapq
def astar(graph,hv,start,goal):
    priority_queue=[(hv[start],0,start,[start])]
    visited={}
    
    while priority_queue:
        f,g,node,path=heapq.heappop(priority_queue)
        if node==goal:
            return path,g

        if node in visited and visited[node]<=g:
            continue
        visited[node]=g

        for n, cost in graph[node]:
            new_g = g+cost
            new_f = new_g+hv[n]
            heapq.heappush(priority_queue,(new_f,new_g, n, path+[n]))
    return None


graph= {
    'A': [('B',1), ('C',4)],
    'B': [('C',2),('D',3)],
    'C': [('E',5)],
    'D': [('G',4),('F',2)],
    'E': [('G',3)],
    'F': [('G',1)],
    'G': []
}

hv = {
    'A': 6,
    'B': 4,
    'C': 3,
    'D': 2,
    'E': 1,
    'F': 1,
    'G': 0
}

path, cost = astar(graph, hv, 'A', 'G')
print("Path:", path) 
print("Cost:", cost)            
            
    
```
